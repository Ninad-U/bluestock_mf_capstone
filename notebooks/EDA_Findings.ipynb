## EDA Key Findings

- **Monthly SIP inflows** showed a steady upward trend from 2022 to 2025, reaching a peak in December 2025.  
  **Chart:** Monthly SIP Inflow Trend (Jan 2022 - Dec 2025)

- **Mutual fund folios nearly doubled** from 13.26 crore in January 2022 to 26.12 crore by December 2025, crossing key milestones during the period.  
  **Chart:** Folio Count Growth with Milestone Annotations

- **SBI Mutual Fund** consistently maintained the highest AUM among all fund houses throughout 2022-2025, growing from ₹6,30,000 crore to ₹12,50,000 crore, demonstrating clear market dominance.  
  **Chart:** AUM Growth by Top 10 Fund Houses (2022 - 2025)

- **Liquid category inflows** remained consistently high, receiving approximately ₹32,000-₹42,000 Cr in monthly inflows, significantly higher than other categories.  
  **Chart:** Category-wise Net Inflows Heatmap

- **Kotak Liquid Fund** recorded the largest absolute NAV increase in the dataset, rising from ₹3,180.63 in January 2022 to ₹4,268.55 by May 2026.  
  **Chart:** NAV Trend Analysis with Annotations (2023 Bull Run, 2024 Corrections)

- **T30 cities contributed 66.3% of SIP inflows**, while B30 cities accounted for 33.7%.  
  **Chart:** State-wise SIP Amount; City Tier Distribution Pie Chart

- **41.1% of SIP investors were aged 26-35**, followed by the 36-45 age group at 24.9%.  
  **Chart:** Investor Age Group Distribution

- **Punjab and Tamil Nadu** recorded the highest total SIP amounts among the analysed states, with other leading states following closely.  
  **Chart:** State-wise SIP Amount

- **SIP ticket sizes remained consistent across age groups**, with median values around ₹5,000 and high-value outliers approaching ₹60,000.  
  **Chart:** SIP Amount Box Plot by Age Group

- **The correlation matrix showed near-zero correlations across the selected fund pairs**, indicating potential diversification benefits among the analysed funds.
